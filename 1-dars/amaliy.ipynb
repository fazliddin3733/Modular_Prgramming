```python
class BankAccount:
    def __init__(self,balance):
        self.balance=balance
    def show_balance(self):
        print(f"Hozirgi balans:{self.balance} so'm")
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} so'm qo'shildi")
    def withdraw(self,amount):
        if amount<=self.balance:
            self.balance-=amount
            print(f"{amount} so'm yechib olindi")
        else:
            print("Hisobda yetarli mablag' mavjud emas!")
account=BankAccount(10000000)
account.show_balance()
account.deposit(5000000)
account.show_balance()
account.withdraw(3000000)
account.show_balance()
        
```

```python
class StudentGrades:
    def __init__(self, students):
        self.students = students

    def highest_grade(self):
        student = max(self.students, key=lambda x: max(x["grades"]))
        grade = max(student["grades"])
        return student["name"], grade

    def lowest_grade(self):
        student = min(self.students, key=lambda x: min(x["grades"]))
        grade = min(student["grades"])
        return student["name"], grade

    def average_grade(self):
        all_grades = []

        for student in self.students:
            all_grades.extend(student["grades"])

        return sum(all_grades) / len(all_grades)
students = [
    {"name": "Ali", "grades": [80, 90, 75]},
    {"name": "Vali", "grades": [95, 85, 100]},
    {"name": "Sardor", "grades": [60, 70, 65]}
]
grades = StudentGrades(students)
print("Eng yuqori ball:", grades.highest_grade())
print("Eng past ball:", grades.lowest_grade())
print("O'rtacha ball:", grades.average_grade())
```
