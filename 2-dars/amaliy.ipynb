```python
class LibraryBook:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        self.available=False
    def show_info(self):
        print(f"Kitob nome:{self.title}")
        print(f"Muallifi:{self.author}")
        if self.available:
            print("Holati:Kutubxonada mavjud")
        else:
            print("Holati:Ijarada")
book1 = LibraryBook("Python Basics", "John Smith")
book1.show_info()
```

```python
class CarPurchase:
    def __init__(self,model,cost):
        self.model=model
        self.cost=cost
        self.available=True
    def show_info(self):
        print(f"Modeli:{self.model}")
        print(f"Narxi:{self.cost}")
        if self.available:
            print("Sotuvda mavjud")
        else:
            print("Sotuvda mavjud emas")
car1=CarPurchase("Cobalt","12 000")
car1.show_info()
```

```python
class CarShop:
    def __init__(self,balance):
        self.balance=balance
        self.cars={
            "Cobalt":160_000_000,
            "Onix":190_000_000,
            "Tracker":220_000_000,
            "Malibu":350_000_000
        }
    def show_cars(self):
        for model,price in self.cars.items():
            print(f"{model}:{price:} so'm")
    def buy_car(self,model):
        if model in self.cars:
            price=self.cars[model]
            if self.balance>=price:
                print("Hisobingiz yetadi sotib olishingiz mumkin!")
            else:
                print("Hisobingiz yetarli emas sotib ololmaysiz")
        else:
            print("Bunday mashina modeli mavjud emas!")
shop=CarShop(200_000_000)
shop.show_cars()
model=input("Moshina nomini kiriting")
shop.buy_car(model)

        
```
